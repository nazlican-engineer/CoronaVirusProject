```python
'''
KİMLİK BİLGİLERİ
iso_code	Ülkenin 3 harfli kodu (TUR, USA gibi)
continent	Kıta 
date	Tarih
'''
```

```python
'''
VAKA & ÖLÜM VERİLERİ
total_cases	O tarihe kadar toplam kümülatif vaka sayısı
new_cases	O günkü yeni vaka sayısı
new_cases_smoothed	Yeni vakaların 7 günlük hareketli ortalaması 
total_deaths / new_deaths / new_deaths_smoothed	Aynı mantık, ölüm için
*_per_million uzantılı olanlar	Yukarıdakilerin milyon kişi başına normalize edilmiş hali — 
ülkeler arası karşılaştırma için bunları kullanacağız
'''
```

```python
'''
SALGININ ŞİDDETİ (REPRODUCTİON & HASTANE)
reproduction_rate	R değeri — bir hastanın ortalama kaç kişiyi enfekte ettiği
(1'in üstü = yayılıyor, altı = yavaşlıyor)
icu_patients / icu_patients_per_million	Yoğun bakımdaki hasta sayısı
hosp_patients / hosp_patients_per_million	Hastanede yatan hasta sayısı
weekly_icu_admissions / weekly_hosp_admissions	Haftalık yeni yoğun bakım/hastane girişleri
(çok az ülke bu veriyi paylaşmış — dolu satır sayısı çok düşük: ~11.000-24.000)
'''
```

```python
'''
TEST VERİLERİ
total_tests / new_tests	Toplam/günlük yapılan test sayısı
positive_rate	Testlerin yüzde kaçı pozitif çıktı
(yüksekse muhtemelen yeterince test yapılmıyor demektir)
tests_per_case	Bir vaka bulmak için kaç test yapıldı
tests_units	Test sayımının birimi (kişi mi, test sayısı mı — bazı ülkeler farklı sayıyor)
'''
```

```python
'''
AŞI VERİLERİ
total_vaccinations	Toplam yapılan doz sayısı (bir kişi 2 doz olduysa 2 sayılır)
people_vaccinated	En az 1 doz olan kişi sayısı
people_fully_vaccinated	Tam aşılı (2 doz vb.) kişi sayısı
total_boosters	Yapılan hatırlatma (booster) dozu sayısı
*_per_hundred uzantılı	Yüz kişi başına oran
'''
```

```python
'''
stringency_index	Hükümetin kısıtlama sertliği endeksi (0-100, kapanma/maske zorunluluğu vb.)
population_density	Nüfus yoğunluğu (km² başına kişi)
median_age	Medyan yaş
aged_65_older / aged_70_older	Nüfusun 65/70 yaş üstü yüzdesi
gdp_per_capita	Kişi başı GSYH
extreme_poverty	Aşırı yoksulluk oranı
cardiovasc_death_rate	Kardiyovasküler hastalık ölüm oranı
diabetes_prevalence	Diyabet yaygınlığı
female_smokers / male_smokers	Sigara içme oranları (cinsiyete göre)
handwashing_facilities	El yıkama imkanına erişim oranı
hospital_beds_per_thousand	Bin kişiye düşen hastane yatağı
life_expectancy	Ortalama yaşam beklentisi
human_development_index	İnsani gelişmişlik endeksi (HDI)
population	Toplam nüfus
'''
```

```python
#Kıta bazlı inceleme yapmak istediğim zaman ayrı bir df daha kullanışlı olur.
continents = ['Africa', 'Asia', 'Europe', 'North America', 
              'South America', 'Oceania']

#gelir grupları için ayrı bir df daha kullanışlı olur.
income_groups = ['High-income countries', 'Low-income countries',
                  'Lower-middle-income countries', 'Upper-middle-income countries']

#Dünya ve AB toplamlarını ayırıyoruz.
world_and_eu = ['World', 'European Union (27)']

#Birleşik Krallık alt bölgelerini ayırıyoruz.
uk_subregions = ['England', 'Scotland', 'Wales', 'Northern Ireland']

# 1. Sadece gerçek, bağımsız ülkeler — ana analizlerimiz için
df_countries = df[~df['location'].isin(continents + income_groups + world_and_eu + uk_subregions)].copy()

# 2. Sadece kıta toplamları
df_continents = df[df['location'].isin(continents)].copy()

# 3. Sadece gelir grubu toplamları
df_income_groups = df[df['location'].isin(income_groups)].copy()

print(f"Ülke bazlı veri: {len(df_countries)} satır")
print(f"Kıta verisi: {len(df_continents)} satır")
print(f"Gelir grubu verisi: {len(df_income_groups)} satır")
```

```python
# 1. Önce raporlama durumu flag'ini oluştur (orijinal boşluğu kaybetmeden)
df_countries['data_reported'] = df_countries['total_cases'].notna()

# 2. Ülke bazında sıralama (ffill için gerekli)
df_countries = df_countries.sort_values(['location', 'date']).copy()

# 3. Analiz için ayrı bir kopya oluştur (orijinal veri saklı kalsın)
df_countries_analysis = df_countries.copy()

# 4. Kümülatif kolonlar için forward fill (yalnızca analiz versiyonunda)
cumulative_cols = ['total_cases', 'total_deaths', 'total_cases_per_million', 'total_deaths_per_million']
df_countries_analysis[cumulative_cols] = df_countries_analysis.groupby('location')[cumulative_cols].ffill()

# 5. Günlük kolonlar için 0 ile doldurma (yalnızca analiz versiyonunda)
daily_cols = ['new_cases', 'new_deaths', 'new_cases_smoothed', 'new_deaths_smoothed']
df_countries_analysis[daily_cols] = df_countries_analysis[daily_cols].fillna(0)

# 6. Verisi girilmeyen ülke/tarih kombinasyonlarını ayrı tut
missing_country_dates = (
    df_countries[df_countries['data_reported'] == False]
    [['location', 'date', 'total_cases', 'new_cases', 'new_deaths']]
    .copy()
)

# 7. Analiz dataframe'i kullanmaya devam et
# Eğer ileride mevcut DF ile çalışacaksan bunu kullan:
# df_countries = df_countries_analysis

# 8. Kontrol
print('Analiz boşluk kontrolü:')
print(df_countries_analysis[cumulative_cols + daily_cols].isnull().sum())
print('\nRaporlanan / raporlanmayan satırlar:')
print(df_countries['data_reported'].value_counts())
print('\nÖrnek eksik veri satırları:')
print(missing_country_dates.head())
```

```python
from scipy import stats
import statsmodels.api as sm

# Standartlaştırma (z-score): her değişkenden ortalamasını çıkar, std sapmasına böl
reg_data_scaled = reg_data.copy()
predictor_cols = ['aged_65_older', 'hospital_beds_per_thousand', 'gdp_per_capita', 'population_density']

for col in predictor_cols:
    reg_data_scaled[col] = stats.zscore(reg_data_scaled[col])

X_scaled = reg_data_scaled[predictor_cols]
X_scaled = sm.add_c onstant(X_scaled)
y_scaled = reg_data_scaled['total_deaths_per_million']  # hedefi olduğu gibi bırakıyoruz

model_scaled = sm.OLS(y_scaled, X_scaled).fit()
print(model_scaled.summary())
```

```python
# Standartlaştırılmış katsayıları ve güven aralıklarını görselleştirelim
#Hangi bağımsız değişkenin ölüm oranına daha güçlü etkisi olduğunu anlamak için standartlaştırılmış katsayıları kullanabiliriz.
coefs = model_scaled.params.drop('const')
conf_int = model_scaled.conf_int().drop('const')
pvalues = model_scaled.pvalues.drop('const')

errors = [(coefs - conf_int[0]), (conf_int[1] - coefs)]
colors = ['crimson' if p < 0.05 else 'gray' for p in pvalues]

plt.figure(figsize=(9, 5))
plt.errorbar(coefs.values, coefs.index, xerr=errors, fmt='o', 
             color='black', ecolor='gray', capsize=5, markersize=0)

for i, (val, color) in enumerate(zip(coefs.values, colors)):
    plt.scatter(val, i, color=color, s=150, zorder=3, edgecolor='black')

plt.axvline(0, color='black', linewidth=0.8, linestyle='--')
plt.yticks(range(len(coefs)), coefs.index)
plt.xlabel('Standartlaştırılmış Katsayı (1 Std. Sapma Değişimin Etkisi)')
plt.title('Standartlaştırılmış Regresyon Katsayıları (Kırmızı = Anlamlı, p<0.05)')
plt.grid(True, axis='x', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```
