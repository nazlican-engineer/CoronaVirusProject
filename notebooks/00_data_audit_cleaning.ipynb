```python
'''
KİMLİK BİLGİLERİ
iso_code	Ülkenin 3 harfli kodu (TUR, USA gibi)
continent	Kıta 
date	Tarih
'''
```

```python
'''
VAKA & ÖLÜM VERİLERİ
total_cases	O tarihe kadar toplam kümülatif vaka sayısı
new_cases	O günkü yeni vaka sayısı
new_cases_smoothed	Yeni vakaların 7 günlük hareketli ortalaması 
total_deaths / new_deaths / new_deaths_smoothed	Aynı mantık, ölüm için
*_per_million uzantılı olanlar	Yukarıdakilerin milyon kişi başına normalize edilmiş hali — 
ülkeler arası karşılaştırma için bunları kullanacağız
'''
```

```python
'''
SALGININ ŞİDDETİ (REPRODUCTİON & HASTANE)
reproduction_rate	R değeri — bir hastanın ortalama kaç kişiyi enfekte ettiği
(1'in üstü = yayılıyor, altı = yavaşlıyor)
icu_patients / icu_patients_per_million	
hosp_patients / hosp_patients_per_million	Hastanede yatan hasta sayısı
weekly_icu_admissions / weekly_hosp_admissions	Haftalık yeni yoğun bakım/hastane girişleri
(çok az ülke bu veriyi paylaşmış — dolu satır sayısı çok düşük: ~11.000-24.000)
'''
```

```python
'''
TEST VERİLERİ
total_tests / new_tests	Toplam/günlük yapılan test sayısı
positive_rate	Testlerin yüzde kaçı pozitif çıktı
(yüksekse muhtemelen yeterince test yapılmıyor demektir)
tests_per_case	Bir vaka bulmak için kaç test yapıldı
tests_units	Test sayımının birimi (kişi mi, test sayısı mı — bazı ülkeler farklı sayıyor)
'''
```

```python
'''
AŞI VERİLERİ
total_vaccinations	Toplam yapılan doz sayısı (bir kişi 2 doz olduysa 2 sayılır)
people_vaccinated	En az 1 doz olan kişi sayısı
people_fully_vaccinated	Tam aşılı (2 doz vb.) kişi sayısı
total_boosters	Yapılan hatırlatma (booster) dozu sayısı
*_per_hundred uzantılı	Yüz kişi başına oran
'''
```

```python
'''
stringency_index	Hükümetin kısıtlama sertliği endeksi (0-100, kapanma/maske zorunluluğu vb.)
population_density	Nüfus yoğunluğu (km² başına kişi)
median_age	Medyan yaş
aged_65_older / aged_70_older	Nüfusun 65/70 yaş üstü yüzdesi
gdp_per_capita	Kişi başı GSYH
extreme_poverty	Aşırı yoksulluk oranı
cardiovasc_death_rate	Kardiyovasküler hastalık ölüm oranı
diabetes_prevalence	Diyabet yaygınlığı
female_smokers / male_smokers	Sigara içme oranları (cinsiyete göre)
handwashing_facilities	El yıkama imkanına erişim oranı
hospital_beds_per_thousand	Bin kişiye düşen hastane yatağı
life_expectancy	Ortalama yaşam beklentisi
human_development_index	İnsani gelişmişlik endeksi (HDI)
population	Toplam nüfus
'''
```

```python
'''
Aşırı Ölüm (Excess Mortality)
excess_mortality	Beklenen ölüm sayısına göre fazladan ölüm yüzdesi
(COVID'in resmi sayılara yansımayan gerçek etkisini gösterir)
excess_mortality_cumulative	Kümülatif hali
excess_mortality_cumulative_absolute	Mutlak sayı olarak
excess_mortality_cumulative_per_million	Milyon kişi başına 
'''
```

```python
#Farklı konum sayısını belirlemek için bunu kullanıyoruz.
print("Farklı konum sayısı:", df['location'].nunique())
print(df['location'].unique())
```

```python
continents = [
    'Africa', 'Asia', 'Europe',
    'North America', 'South America', 'Oceania'
]

df_continents = df[df['location'].isin(continents)].copy()

print(df_continents['location'].unique())
print("Kıta toplamlarına ait satır sayısı:", len(df_continents))
```

```python
continents = ['Africa', 'Asia', 'Europe', 'North America', 
              'South America', 'Oceania']

income_groups = ['High-income countries', 'Low-income countries',
                  'Lower-middle-income countries', 'Upper-middle-income countries']

world_and_eu = ['World', 'European Union (27)']

uk_subregions = ['England', 'Scotland', 'Wales', 'Northern Ireland']

# 1. Sadece gerçek, bağımsız ülkeler — ana analizlerimiz için
df_countries = df[~df['location'].isin(continents + income_groups + world_and_eu + uk_subregions)].copy()

# 2. Sadece kıta toplamları
df_continents = df[df['location'].isin(continents)].copy()

# 3. Sadece gelir grubu toplamları
df_income_groups = df[df['location'].isin(income_groups)].copy()

print(f"Ülke bazlı veri: {len(df_countries)} satır")
print(f"Kıta verisi: {len(df_continents)} satır")
print(f"Gelir grubu verisi: {len(df_income_groups)} satır")
```

```python
#Kalan konumların kıta bilgisi kaldı mı 
missing_continent = df_countries.loc[
    df_countries['continent'].isna(), 'location'
].unique()

print("Kıta bilgisi eksik konumlar:", missing_continent)
```

```python
location_counts = (
    df_countries.groupby('location')
    .size()
    .sort_values()
)

print("En az kaydı olan 10 konum:")
print(location_counts.head(10))

print("\nEn fazla kaydı olan 10 konum:")
print(location_counts.tail(10))
```

```python
#Aynı tarihlerden değerler var mı?
location_summary = df_countries.groupby('location').agg(
    satir_sayisi=('date', 'size'),
    farkli_tarih_sayisi=('date', 'nunique')
)

location_summary['tekrar_sayisi'] = (
    location_summary['satir_sayisi']
    - location_summary['farkli_tarih_sayisi']
)

print(
    location_summary.sort_values(
        'tekrar_sayisi', ascending=False
    ).head(10)
)
```

```python
east_timor = df_countries[
    df_countries['location'] == 'East Timor'
].copy()

east_timor_duplicates = east_timor[
    east_timor.duplicated(subset=['date'], keep=False)
].sort_values('date')

# İlk tekrarlanan tarihin kayıtlarını yan yana göster
first_date = east_timor_duplicates['date'].iloc[0]

display(
    east_timor_duplicates[
        east_timor_duplicates['date'] == first_date
    ].T
)
```

```python
#ekrarlayan her tarihten kaç tane var
date_counts = east_timor.groupby('date').size()

print(date_counts[date_counts > 1].sort_index())
```

```python
#Aynı tarie ait satırlarda,aynı sütunda birbirinden farklı değerler var mı?
unique_counts = (
    east_timor_duplicates
    .groupby('date')
    .nunique(dropna=True)
)

conflicts = unique_counts.gt(1)

print("Çelişki bulunan sütunlar ve tarih sayıları:")
print(conflicts.sum()[conflicts.sum() > 0])

print("En az bir çelişki bulunan tarih sayısı:")
print(conflicts.any(axis=1).sum())
```

```python
east_timor_clean = (
    east_timor
    .sort_values('date')
    .groupby('date', as_index=False)
    .first()
)

print("Önceki satır sayısı:", len(east_timor))
print("Birleştirme sonrası:", len(east_timor_clean))
print("Tekrarlanan tarih:", east_timor_clean['date'].duplicated().sum())
```

```python
df_countries = pd.concat(
    [
        df_countries[df_countries['location'] != 'East Timor'],
        east_timor_clean
    ],
    ignore_index=True
)
```

```python
#Faroe Islands için de aynı incelemeyi yapacağız
faroe_islands = df_countries[
    df_countries['location'] == 'Faroe Islands'
].copy()

faroe_islands_duplicates = faroe_islands[
    faroe_islands.duplicated(subset=['date'], keep=False)
].sort_values('date')

# İlk tekrarlanan tarihin kayıtlarını yan yana göster
first_date = faroe_islands_duplicates['date'].iloc[0]

display(
    faroe_islands_duplicates[
        faroe_islands_duplicates['date'] == first_date
    ].T
)
```

```python
#Aynı tarie ait satırlarda,aynı sütunda birbirinden farklı değerler var mı?
unique_counts = (
    faroe_islands_duplicates
    .groupby('date')
    .nunique(dropna=True)
)

conflicts = unique_counts.gt(1)

print("Çelişki bulunan sütunlar ve tarih sayıları:")
print(conflicts.sum()[conflicts.sum() > 0])

print("En az bir çelişki bulunan tarih sayısı:")
print(conflicts.any(axis=1).sum())
```

```python
faroe_islands_clean = (
    faroe_islands
    .sort_values('date')
    .groupby('date', as_index=False)
    .first()
)

print("Önceki satır sayısı:", len(faroe_islands))
print("Birleştirme sonrası:", len(faroe_islands_clean))
print("Tekrarlanan tarih:", faroe_islands_clean['date'].duplicated().sum())
```

```python
df_countries = pd.concat(
    [
        df_countries[df_countries['location'] != 'Faroe Islands'],
        faroe_islands_clean
    ],
    ignore_index=True
)
```

```python
print(
    "Kalan ülke–tarih tekrarı:",
    df_countries.duplicated(subset=['location', 'date']).sum()
)
```

```python
#Aynı tarihlerden değerler var mı?
location_summary = df_countries.groupby('location').agg(
    satir_sayisi=('date', 'size'),
    farkli_tarih_sayisi=('date', 'nunique')
)

location_summary['tekrar_sayisi'] = (
    location_summary['satir_sayisi']
    - location_summary['farkli_tarih_sayisi']
)

print(
    location_summary.sort_values(
        'tekrar_sayisi', ascending=False
    ).head(10)
)
#tekrarlayan herhangi bir değer kalmadı
```

```python
#hangi tarih aralıklarında veriler var ama bu tarihler arasında boş olan değerleri göstermiyor
df_countries['date'] = pd.to_datetime(df_countries['date'])

date_coverage = df_countries.groupby('location').agg(
    ilk_tarih=('date', 'min'),
    son_tarih=('date', 'max'),
    kayitli_gun=('date', 'nunique')
)

print(date_coverage.sort_values('kayitli_gun').head(10))
```

```python
#Belirtilen tarih aralıklarında kaç günün verisi yok
date_coverage['beklenen_gun'] = (
    date_coverage['son_tarih'] - date_coverage['ilk_tarih']
).dt.days + 1

date_coverage['eksik_tarih_sayisi'] = (
    date_coverage['beklenen_gun'] - date_coverage['kayitli_gun']
)

print(
    date_coverage.sort_values(
        'eksik_tarih_sayisi', ascending=False
    ).head(10)
)
#sadece bir tarih girilmemiş
```
