```python
# Kıta toplamlarının yer aldığı konumlar
continents = [
    'Africa', 'Asia', 'Europe',
    'North America', 'South America', 'Oceania'
]

# Gelir seviyesine göre oluşturulmuş ülke grupları
income_groups = [
    'High-income countries',
    'Low-income countries',
    'Lower-middle-income countries',
    'Upper-middle-income countries'
]

# Dünya ve Avrupa Birliği toplamları
world_and_eu = [
    'World',
    'European Union (27)'
]

# Ana ülke analizinden ayrı tutulacak bölgeler,
# bağlı bölgeler ve özel statülü konumlar
region_locations = [
    'American Samoa', 'Anguilla', 'Aruba', 'Bermuda',
    'Bonaire Sint Eustatius and Saba', 'British Virgin Islands',
    'Cayman Islands', 'Cook Islands', 'Curacao', 'Faroe Islands',
    'Falkland Islands', 'French Guiana', 'French Polynesia',
    'Gibraltar', 'Greenland', 'Guadeloupe', 'Guam', 'Guernsey',
    'Hong Kong', 'Isle of Man', 'Jersey', 'Macao', 'Martinique',
    'Mayotte', 'Montserrat', 'New Caledonia', 'Niue',
    'Northern Cyprus', 'Northern Mariana Islands', 'Palestine',
    'Pitcairn', 'Puerto Rico', 'Reunion', 'Saint Barthelemy',
    'Saint Helena', 'Saint Martin (French part)',
    'Saint Pierre and Miquelon', 'Sint Maarten (Dutch part)',
    'Taiwan', 'Tokelau', 'Turks and Caicos Islands',
    'United States Virgin Islands', 'Wallis and Futuna',
    'Western Sahara', 'Kosovo',

    # Birleşik Krallık alt bölgeleri
    'England', 'Scotland', 'Wales', 'Northern Ireland'
]

# Ülke analizine dahil edilmeyecek bütün konumları birleştiriyoruz
excluded_locations = (
    continents
    + income_groups
    + world_and_eu
    + region_locations
)

# Yalnızca ana ülke analizinde kullanılacak satırlar
df_countries = df[
    ~df['location'].isin(excluded_locations)
].copy()

# Bölgeler, bağlı bölgeler ve özel statülü konumlara ait satırlar
df_regions = df[
    df['location'].isin(region_locations)
].copy()

# Kıta toplamlarına ait satırlar
df_continents = df[
    df['location'].isin(continents)
].copy()

# Gelir grubu toplamlarına ait satırlar
df_income_groups = df[
    df['location'].isin(income_groups)
].copy()

# Dünya ve Avrupa Birliği toplamlarına ait satırlar
df_world_and_eu = df[
    df['location'].isin(world_and_eu)
].copy()

# Oluşturduğumuz DataFrame'lerin satır sayılarını kontrol ediyoruz
print("Ülke satır sayısı:", len(df_countries))
print("Bölge satır sayısı:", len(df_regions))
print("Kıta satır sayısı:", len(df_continents))
print("Gelir grubu satır sayısı:", len(df_income_groups))
print("Dünya ve AB toplamı satır sayısı:", len(df_world_and_eu))

# Bölge olarak ayrılan konumların isimlerini gösteriyoruz
print("\nBölge olarak ayrılan konumlar:")
print(df_regions['location'].unique())
```

```python
east_timor = df_countries[
    df_countries['location'] == 'East Timor'
].copy()

east_timor_duplicates = east_timor[
    east_timor.duplicated(subset=['date'], keep=False)
].sort_values('date')
#subset:aynı tarihten değerler var mı?

# İlk tekrarlanan tarihin kayıtlarını yan yana göster
first_date = east_timor_duplicates['date'].iloc[0]
#iloc:listenin ilk tarihindeki satırı getir.

display(
    east_timor_duplicates[
        east_timor_duplicates['date'] == first_date
    ].T
)
```

```python
# Karşılaştırmada konum ve tarih sütunlarını dışarıda bırakıyoruz
columns_to_compare = [
    column
    for column in east_timor_duplicates.columns
    if column not in ['location', 'date']
]

# Her tarih ve sütun için kaç farklı dolu değer bulunduğunu hesaplıyoruz
unique_value_counts = (
    east_timor_duplicates
    .groupby('date')[columns_to_compare]
    .nunique(dropna=True)
)

# Bir sütunda birden fazla farklı dolu değer varsa çelişki vardır
conflict_dates = unique_value_counts[
    unique_value_counts.gt(1).any(axis=1)
]

print("Tekrar eden tarih sayısı:", east_timor_duplicates['date'].nunique())
print("Çelişkili tarih sayısı:", len(conflict_dates))

print("\nİlk çelişkili tarihler:")
print(conflict_dates.head())
#sonuç olarak  birbirine benzeyen tarihlerde iki farklı değer yok birisi doluysa diğeri boş yüksek ihtimal
```

```python
#bütün parçaladığım kıtalar,bölgeler(ve diğerleri için)  duplicate değer olup olmadığını kontrol eder
dataframes = {
    'regions': df_regions,
    'continents': df_continents,
    'income_groups': df_income_groups,
    'world_and_eu': df_world_and_eu
}

duplicate_results = []

for dataframe_name, dataframe in dataframes.items():

    summary = (
        dataframe
        .groupby('location')
        .agg(
            satir_sayisi=('date', 'size'),
            farkli_tarih_sayisi=('date', 'nunique')
        )
    )

    summary['tekrar_sayisi'] = (
        summary['satir_sayisi']
        - summary['farkli_tarih_sayisi']
    )

    summary = summary[
        summary['tekrar_sayisi'] > 0
    ].copy()

    summary['veri_grubu'] = dataframe_name

    duplicate_results.append(
        summary.reset_index()
    )

duplicate_audit = pd.concat(
    duplicate_results,
    ignore_index=True
)

print(
    duplicate_audit.sort_values(
        'tekrar_sayisi',
        ascending=False
    )
)
```

```python
conflict_results = []

for dataframe_name, dataframe in dataframes.items():

    duplicate_rows = dataframe[
        dataframe.duplicated(
            subset=['location', 'date'],
            keep=False
        )
    ].copy()

    if duplicate_rows.empty:
        continue

    columns_to_compare = [
        column
        for column in duplicate_rows.columns
        if column not in ['location', 'date']
    ]

    unique_value_counts = (
        duplicate_rows
        .groupby(['location', 'date'])[columns_to_compare]
        .nunique(dropna=True)
    )

    conflict_by_date = unique_value_counts.gt(1).any(axis=1)

    group_summary = pd.DataFrame({
        'tekrar_eden_tarih_sayisi': (
            duplicate_rows
            .groupby('location')['date']
            .nunique()
        ),
        'celiskili_tarih_sayisi': (
            conflict_by_date
            .groupby(level='location')
            .sum()
        )
    }).reset_index()

    group_summary['veri_grubu'] = dataframe_name

    conflict_results.append(group_summary)

conflict_audit = pd.concat(
    conflict_results,
    ignore_index=True
)

print(
    conflict_audit.sort_values(
        'celiskili_tarih_sayisi',
        ascending=False
    )
)
```

```python
def merge_duplicate_dates(dataframe):
    return (
        dataframe
        .sort_values(['location', 'date'])
        .groupby(
            ['location', 'date'],
            as_index=False
        )
        .first()
    )

df_regions = merge_duplicate_dates(df_regions)

df_income_groups = merge_duplicate_dates(
    df_income_groups
)

df_world_and_eu = merge_duplicate_dates(
    df_world_and_eu
)
```

```python
dataframes = {
    'regions': df_regions,
    'continents': df_continents,
    'income_groups': df_income_groups,
    'world_and_eu': df_world_and_eu
}

for dataframe_name, dataframe in dataframes.items():
    duplicate_count = dataframe.duplicated(
        subset=['location', 'date']
    ).sum()

    print(
        dataframe_name,
        "kalan tekrar:",
        duplicate_count
    )
```

```python
#Ülkelerin,bölgelerin,kıtaların kayıtlı ilk tarih ve son tarihlerine bakıyoruz ve o sütuna ait toplam kaç satır var inceliyoruz.
dataframes = {
    'countries': df_countries,
    'regions': df_regions,
    'continents': df_continents,
    'income_groups': df_income_groups,
    'world_and_eu': df_world_and_eu
}

coverage_results = []

for dataframe_name, dataframe in dataframes.items():

    dataframe['date'] = pd.to_datetime(
        dataframe['date']
    )

    coverage = (
        dataframe
        .groupby('location')
        .agg(
            ilk_tarih=('date', 'min'),
            son_tarih=('date', 'max'),
            kayitli_gun=('date', 'nunique')
        )
    )

    coverage['veri_grubu'] = dataframe_name

    coverage_results.append(
        coverage.reset_index()
    )

date_coverage_all = pd.concat(
    coverage_results,
    ignore_index=True
)

print(
    date_coverage_all
    .sort_values('kayitli_gun')
    .head(100)
)
```

```python
#ilk ve son tarih arasında eksik gün var mı onu tespit ediyoruz.
date_coverage_all['beklenen_gun'] = (
    (
        date_coverage_all['son_tarih']
        - date_coverage_all['ilk_tarih']
    )
    .dt.days
    + 1
)

date_coverage_all['eksik_gun'] = (
    date_coverage_all['beklenen_gun']
    - date_coverage_all['kayitli_gun']
)

print(
    date_coverage_all[
        date_coverage_all['eksik_gun'] > 0
    ]
    .sort_values('eksik_gun', ascending=False)
)
```

```python
#eksik gün değeri sıfırdan büyük olan konumları seçer
gap_locations = date_coverage_all[
    date_coverage_all['eksik_gun'] > 0
].copy()

missing_date_results = []

for _, row in gap_locations.iterrows():

    dataframe = dataframes[row['veri_grubu']]

    observed_dates = pd.DatetimeIndex(
        dataframe.loc[
            dataframe['location'] == row['location'],
            'date'
        ].unique()
    )

    expected_dates = pd.date_range(
        start=row['ilk_tarih'],
        end=row['son_tarih'],
        freq='D'
    )

    missing_dates = expected_dates.difference(
        observed_dates
    )

    missing_date_results.append(
        pd.DataFrame({
            'veri_grubu': row['veri_grubu'],
            'location': row['location'],
            'eksik_tarih': missing_dates
        })
    )

missing_dates_all = pd.concat(
    missing_date_results,
    ignore_index=True
)

print(missing_dates_all)
#sonuç olarak 2022-12-05 tarihinde regions bölgesinde location'ı Northern Cyprus olan bölgede veri yok bunu dashboardda ayrıca belirtmek gerekebilir
```

## Veri Denetimi Sonuçları

- Veri setindeki konumlar; ana analiz ülkeleri, bölgeler/özel statülü konumlar, kıtalar, gelir grupları ile Dünya ve Avrupa Birliği toplamları olarak ayrıldı.
- Western Sahara, Faroe Islands ve Birleşik Krallık alt bölgeleri gibi ülke olarak değerlendirilmemesi gereken konumlar `df_regions` içinde tutuldu.

- East Timor, Faroe Islands, gelir grupları ve Avrupa Birliği toplamlarında tekrar eden tarihli kayıtlar bulundu.

- Çelişkili değer bulunmadığı için tekrar eden kayıtlar, her sütundaki boş olmayan değer korunarak tekilleştirildi.
- Tekilleştirme sonrasında tüm veri gruplarında kalan `location + date` tekrarı kontrol edildi.
- Her konumun ilk ve son kayıt tarihi, kayıtlı gün sayısı ve bu aralıktaki eksik gün sayısı hesaplandı.

- Eksik tarih, sıfır vaka veya sıfır ölüm olarak yorumlanmadı; veri kalitesi notu olarak korundu.
